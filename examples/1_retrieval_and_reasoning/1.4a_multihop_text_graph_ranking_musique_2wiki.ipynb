# 1.4a Scaled Multi-Hop QA from Text: Vector RAG vs Graph RAG on MuSiQue and 2WikiMultiHopQA

*Formerly notebook 11.*

**Presentation map:** Slide 32 (vector RAG is the baseline to beat), Slide 33 (graph construction from text),
Slide 36 (graphs help on multi-hop questions), Slide 38 (decision rule by question type), Slide 41 (multi-hop
traversal).

Notebook 8 showed a large graph advantage on MetaQA, but MetaQA *hands the system a clean, complete knowledge graph*
and asks templated questions — the most graph-friendly setting possible. This notebook removes that advantage: the
graph must be **built from raw Wikipedia paragraphs by the LLM**, so every extraction error counts against it.

**Benchmarks** (the standard HippoRAG / HippoRAG 2 evaluation sets, Hugging Face `osunlp/HippoRAG_2`):

| Benchmark | Questions | Passage pool | What makes it hard |
|---|---|---|---|
| **MuSiQue** (answerable, dev) | 1,000 (2-, 3- and 4-hop) | 11,656 paragraphs | questions composed so that no single paragraph answers them; many distractors |
| **2WikiMultiHopQA** (dev) | 1,000 (compositional, comparison, inference, bridge-comparison) | 6,119 paragraphs | two-entity chains built from Wikidata relations |

**Two retrievers, one answer prompt, same budget of 5 passages:**
- **(a) Vector RAG** — `text-embedding-3-small` over every passage, top-5 by cosine similarity.
- **(b) Graph RAG (HippoRAG-2-style)** — the LLM extracts entities and subject–predicate–object triples from every
  passage (one structured call per passage); entities with the same normalised name are merged, and near-duplicate
  names (embedding similarity ≥ 0.9) get a synonym edge. At query time the question is matched against the triples,
  the entities of the best-matching triples seed a **personalised PageRank** (damping 0.5, plus a small weight on the
  dense passage scores, as in HippoRAG 2), and the 5 highest-ranked passages go to the answerer.

**Scoring:** exact match (EM) and token F1 against the gold answer and its aliases — the standard metrics of both
benchmarks, computed in code (no LLM judge). Retrieval quality is measured as **recall@5**: the share of the gold
supporting paragraphs among the 5 retrieved passages.

**Model:** `gpt-6-luna`, no reasoning, temperature 0, through the OpenAI **Batch API** (helper `smash_batch.py`).
Everything is cached in `data/benchmarks/musique_2wiki/cache/`; the default `SMASH_MODE=replay` re-runs the saved
results in about a minute without calling the API.

```python
# Notebooks live in examples/<task group>/; data/, smash_batch.py and the .venv-* environments stay in examples/.
# Every path below is relative to examples/, so run from there whether the notebook is opened in its own folder or not.
import os, sys
if not os.path.exists("smash_batch.py") and os.path.exists(os.path.join("..", "smash_batch.py")):
    os.chdir("..")
if os.getcwd() not in sys.path:
    sys.path.insert(0, os.getcwd())
```

```python
import json, os, re, string, time, random
from collections import Counter, defaultdict
import numpy as np
import scipy.sparse as sp
from pydantic import BaseModel, Field
from smash_batch import LLM, mode_from_env, wilson, paired_boot

MODE = mode_from_env()               # replay (default) | test (small live run) | batch (full run via Batch API)
TEST = MODE == "test"
BASE = "data/benchmarks/musique_2wiki"
llm = LLM(f"{BASE}/cache" + ("_test" if TEST else ""))
N_Q = 25 if TEST else 1000           # questions per benchmark (test mode: the first 25 of the same fixed order)

def load(name):
    qs = json.load(open(f"{BASE}/{name}.json"))
    corpus = json.load(open(f"{BASE}/{name}_corpus.json"))
    return qs, corpus

DATA = {}
for name in ("musique", "2wikimultihopqa"):
    qs, corpus = load(name)
    for q in qs:                                     # one schema for both benchmarks
        if name == "musique":
            q["qid"], q["golds"] = q["id"], [q["answer"]] + q.get("answer_aliases", [])
            q["support"] = {p["title"] + "\n" + p["paragraph_text"] for p in q["paragraphs"] if p["is_supporting"]}
            q["group"] = q["id"].split("hop")[0] + "-hop"
        else:
            q["qid"], q["golds"] = q["_id"], [q["answer"]]
            ctx = {t: " ".join(s) for t, s in q["context"]}
            q["support"] = {t + "\n" + ctx[t] for t in {t for t, _ in q["supporting_facts"]} if t in ctx}
            q["group"] = q["type"]
    DATA[name] = {"questions": qs[:N_Q], "corpus": corpus}
    print(f"{name}: {len(qs):,} questions, {len(corpus):,} passages | groups:",
          dict(Counter(q["group"] for q in qs)))
print("MODE =", MODE)
```

## The passage pools

Each benchmark is indexed as one pool. In test mode the pool is cut down to the passages of the test questions, so
the whole pipeline can be checked for a few cents.

```python
for name, d in DATA.items():
    passages = [p["title"] + "\n" + p["text"] for p in d["corpus"]]
    if TEST:
        keep = set()
        for q in d["questions"]:
            if name == "musique":
                keep |= {p["title"] + "\n" + p["paragraph_text"] for p in q["paragraphs"]}
            else:
                keep |= {t + "\n" + " ".join(s) for t, s in q["context"]}
        passages = [p for p in passages if p in keep]
    d["passages"] = passages
    d["pid"] = {p: i for i, p in enumerate(passages)}
    found = np.mean([len(q["support"] & set(d["pid"])) / max(len(q["support"]), 1) for q in d["questions"]])
    print(f"{name}: {len(passages):,} passages in the pool; gold supporting paragraphs found in the pool: {found:.1%}")
```

## Index (a): dense embeddings — the vector baseline's whole indexing cost

```python
for name, d in DATA.items():
    d["P"] = llm.embed(f"{name}_passages", d["passages"])
    d["Q"] = llm.embed(f"{name}_questions", [q["question"] for q in d["questions"]])
    print(f"{name}: passage matrix {d['P'].shape}, question matrix {d['Q'].shape}")
```

## Index (b): LLM triple extraction → entity–passage graph (Slide 33)

One structured-output request per passage (OpenIE, as in HippoRAG). All requests of a benchmark go out as one
Batch API job. Every entity keeps the ids of the passages it came from (provenance), so the graph can always point
back to its source text.

```python
class Triple(BaseModel):
    subject: str
    predicate: str
    object: str

class OpenIE(BaseModel):
    entities: list[str] = Field(description="named entities, dates and numbers mentioned in the passage")
    triples: list[Triple] = Field(description="factual subject-predicate-object triples stated in the passage")

OPENIE_SYS = ("Extract an open knowledge graph from a Wikipedia passage. List the named entities (people, places, "
              "organisations, works, events, dates, numbers) and every factual triple the passage states. Use full, "
              "canonical entity names (resolve pronouns to the entity they refer to); keep dates and numbers exact. At most "
              "40 triples, most important first.")

TRIPLE_RE = re.compile(r'\{\s*"subject":\s*"((?:[^"\\]|\\.)*)",\s*"predicate":\s*"((?:[^"\\]|\\.)*)",\s*"object":\s*"((?:[^"\\]|\\.)*)"\s*\}')

def salvage(text):
    # An output cut off at the token limit still contains complete triples: keep those.
    return {"entities": [], "triples": [{"subject": a, "predicate": b, "object": c} for a, b, c in TRIPLE_RE.findall(text)]}

for name, d in DATA.items():
    reqs = {f"p{i}": LLM.body([{"role": "system", "content": OPENIE_SYS}, {"role": "user", "content": p}],
                              max_tokens=4000, schema=OpenIE)
            for i, p in enumerate(d["passages"])}
    res = llm.run(f"{name}_openie", reqs, parse=OpenIE)
    d["openie"] = [res[f"p{i}"]["parsed"] or salvage(res[f"p{i}"].get("text", "")) for i in range(len(d["passages"]))]
    bad = sum(res[f"p{i}"]["parsed"] is None for i in range(len(d["passages"])))
    nt = sum(len(x["triples"]) for x in d["openie"])
    print(f"{name}: {len(reqs):,} extraction requests, {bad} cut off (complete triples kept), {nt:,} triples "
          f"({nt / len(reqs):.1f} per passage)")
```

```python
norm = lambda s: re.sub(r"\s+", " ", re.sub(r"[^a-z0-9 ]", " ", str(s).lower())).strip()

def build_graph(d):
    # Nodes: passages 0..P-1, then entities. Edges: entity-entity (triple), entity-passage (mention), synonyms.
    n_p = len(d["passages"])
    ent_id, ent_name = {}, []
    def eid(s):
        k = norm(s)
        if not k:
            return None
        if k not in ent_id:
            ent_id[k] = n_p + len(ent_name); ent_name.append(s)
        return ent_id[k]
    rows, cols, facts = [], [], []
    for pi, ex in enumerate(d["openie"]):
        for e in ex["entities"]:
            j = eid(e)
            if j is not None:
                rows += [pi, j]; cols += [j, pi]
        for t in ex["triples"]:
            a, b = eid(t["subject"]), eid(t["object"])
            if a is None or b is None or a == b:
                continue
            rows += [a, b, pi, a, pi, b]; cols += [b, a, a, pi, b, pi]
            facts.append((a, b, pi, f"{t['subject']} {t['predicate']} {t['object']}"))
    d["n_p"], d["ent_name"], d["facts"] = n_p, ent_name, facts
    # Synonym edges between near-duplicate entity names (HippoRAG 2), computed on entity-name embeddings.
    E = llm.embed(f"{d['name']}_entities", ent_name, cap=200)
    syn = 0
    for s in range(0, len(ent_name), 512):
        S = E[s:s + 512] @ E.T
        ii, jj = np.where(S >= 0.9)
        for a, b in zip(ii + s, jj):
            if a < b:
                rows += [n_p + a, n_p + b]; cols += [n_p + b, n_p + a]; syn += 1
    n = n_p + len(ent_name)
    A = sp.csr_matrix((np.ones(len(rows), np.float32), (rows, cols)), shape=(n, n))
    A.data[:] = 1.0                                  # duplicate edges count once
    deg = np.asarray(A.sum(1)).ravel(); deg[deg == 0] = 1
    d["T"] = (sp.diags(1 / deg) @ A).T.tocsr()        # column-stochastic transition matrix
    d["F"] = llm.embed(f"{d['name']}_facts", [f[3] for f in facts], cap=300)
    return n, syn

for name, d in DATA.items():
    d["name"] = name
    n, syn = build_graph(d)
    print(f"{name}: graph with {d['n_p']:,} passage + {len(d['ent_name']):,} entity nodes, "
          f"{len(d['facts']):,} triples, {syn:,} synonym edges")
```

## Two retrievers with the same budget of 5 passages

**Vector:** the 5 passages most similar to the question. **Graph:** the question embedding is compared with every
triple; the entities of the 5 best triples (weighted by similarity) plus 0.05 × the dense score of the 20 best
passages form the personalisation vector of a personalised PageRank (damping 0.5); the 5 passages with the highest
PageRank go to the answerer. Recall@5 counts how many gold supporting paragraphs each retriever found.

```python
K = 5

def ppr(T, pers, alpha=0.5, iters=40):
    p = pers / pers.sum(); x = p.copy()
    for _ in range(iters):
        x = (1 - alpha) * p + alpha * (T @ x)
    return x

def retrieve(d, qi):
    q = d["Q"][qi]
    dense = d["P"] @ q
    vec = list(np.argsort(-dense)[:K])
    pers = np.zeros(d["T"].shape[0], np.float64)
    fs = d["F"] @ q if len(d["facts"]) else np.zeros(0)
    for j in np.argsort(-fs)[:5]:
        a, b, pi, _ = d["facts"][j]
        pers[a] += fs[j]; pers[b] += fs[j]
    for j in np.argsort(-dense)[:20]:
        pers[j] += 0.05 * max(dense[j], 0)
    if pers.sum() == 0:
        return vec, vec
    x = ppr(d["T"], pers)[:d["n_p"]]
    return vec, list(np.argsort(-x)[:K])

def recall(d, q, ids):
    if not q["support"]:
        return float("nan")
    got = {d["passages"][i] for i in ids}
    return len(q["support"] & got) / len(q["support"])

t0 = time.time()
for name, d in DATA.items():
    d["ret"] = [retrieve(d, i) for i in range(len(d["questions"]))]
    rv = np.nanmean([recall(d, q, v) for q, (v, g) in zip(d["questions"], d["ret"])])
    rg = np.nanmean([recall(d, q, g) for q, (v, g) in zip(d["questions"], d["ret"])])
    print(f"{name}: recall@5 of gold supporting paragraphs  vector {rv:.1%}  |  graph {rg:.1%}")
print(f"retrieval for {sum(len(d['questions']) for d in DATA.values()):,} questions in {time.time() - t0:.0f} s")
```

## Answers (one Batch job) and EM / F1 scoring

```python
ANSWER_SYS = ("Answer the question using the passages. Reason step by step in at most three short sentences, then "
              "give the final answer on the last line as 'Answer: <answer>'. The answer must be a short phrase "
              "(an entity, date, number or yes/no), not a sentence.")

def ctx(d, ids):
    return "\n\n".join(f"[{n + 1}] {d['passages'][i]}" for n, i in enumerate(ids))

def final(text):
    m = re.findall(r"Answer:\s*(.+)", text)
    return (m[-1] if m else text.strip().splitlines()[-1] if text.strip() else "").strip()

def normalize(s):
    s = s.lower()
    s = "".join(ch for ch in s if ch not in set(string.punctuation))
    s = re.sub(r"\b(a|an|the)\b", " ", s)
    return " ".join(s.split())

def f1(pred, gold):
    p, g = normalize(pred).split(), normalize(gold).split()
    common = Counter(p) & Counter(g); same = sum(common.values())
    if same == 0:
        return 0.0
    pr, rc = same / len(p), same / len(g)
    return 2 * pr * rc / (pr + rc)

for name, d in DATA.items():
    reqs = {}
    for qi, q in enumerate(d["questions"]):
        for m, ids in zip(("vector", "graph"), d["ret"][qi]):
            reqs[f"{m}|{q['qid']}"] = LLM.body([{"role": "system", "content": ANSWER_SYS},
                                               {"role": "user", "content": f"{ctx(d, ids)}\n\nQuestion: {q['question']}"}],
                                              max_tokens=300)
    res = llm.run(f"{name}_answers", reqs)
    d["scores"] = []
    for qi, q in enumerate(d["questions"]):
        row = {"group": q["group"]}
        for m, ids in zip(("vector", "graph"), d["ret"][qi]):
            pred = final(res[f"{m}|{q['qid']}"]["text"])
            row[m] = {"pred": pred, "em": float(max(normalize(pred) == normalize(g) for g in q["golds"])),
                      "f1": max(f1(pred, g) for g in q["golds"]), "recall": recall(d, q, ids)}
        d["scores"].append(row)
    print(f"{name}: {len(reqs):,} answers")
```

## Results with 95% confidence intervals

EM intervals are Wilson intervals; the *graph − vector* column is a paired bootstrap over questions on F1. An
interval that excludes 0 is a difference this sample can resolve.

```python
SUMMARY = {}
def table(name, d):
    print(f"\n{name}  (n = {len(d['scores'])})")
    print(f"{'group':<20}{'n':>5}{'EM vector':>18}{'EM graph':>18}{'F1 vector':>11}{'F1 graph':>10}{'F1 graph-vector [95% CI]':>28}"
          f"{'recall@5 v/g':>15}")
    groups = sorted({s["group"] for s in d["scores"]}) + ["ALL"]
    for g in groups:
        rows = [s for s in d["scores"] if g == "ALL" or s["group"] == g]
        n = len(rows)
        out = {}
        for m in ("vector", "graph"):
            k = int(sum(r[m]["em"] for r in rows)); lo, hi = wilson(k, n)
            out[m] = (k / n, lo, hi, np.mean([r[m]["f1"] for r in rows]), np.nanmean([r[m]["recall"] for r in rows]))
        dm, dlo, dhi = paired_boot([r["vector"]["f1"] for r in rows], [r["graph"]["f1"] for r in rows])
        SUMMARY[(name, g)] = (n, out, (dm, dlo, dhi))
        em = lambda t: f"{t[0]:.1%} [{t[1]:.0%}-{t[2]:.0%}]"
        print(f"{g:<20}{n:>5}{em(out['vector']):>18}{em(out['graph']):>18}{out['vector'][3]:>11.1%}{out['graph'][3]:>10.1%}"
              f"{f'{dm:+.1%} [{dlo:+.1%}, {dhi:+.1%}]':>28}{f'{out['vector'][4]:.0%}/{out['graph'][4]:.0%}':>15}")

for name, d in DATA.items():
    table(name, d)
```

## Amendment (2026-11-08): hybrid retrieval with LLM rerank, on both sides

*Added on 2026-11-08, after the confidence run, as a sensitivity analysis (`baselines.md`, item 4). Written down
here before it runs. The original primary result above (graph − vector, top-5 dense, no rerank) stands and is not
replaced; the verdict line will name both.*

**Why.** The vector baseline above is `text-embedding-3-small` top 5 with no reranking. Published baselines are
stronger (HippoRAG 2 uses a 7B embedder; GraphRAG-Bench's baseline is "RAG with rerank"). The question is whether the
graph's gain survives a stronger retriever, and whether the graph still adds something when it gets the same
improvement.

**Systems** (same 2,000 questions, same passage pools, same answer prompt `ANSWER_SYS`, same 5-passage budget):
- **Hybrid + rerank (vector side).** BM25 (Okapi, k1 = 1.5, b = 0.75, lowercase `\w+` tokens, idf
  log(1 + (N − n + 0.5)/(n + 0.5))) and the existing dense scores (`text-embedding-3-small`, the embeddings
  already cached above) are fused by reciprocal rank fusion (score = Σ 1/(60 + rank)); the top 50 fused passages
  are the candidates.
- **Graph + rerank.** The graph retriever above unchanged (same personalised PageRank), but its **top 50** passages
  by PageRank score instead of its top 5. (The code checks that the first 5 equal the original graph retrieval.)
- **The same LLM reranker for both** (prompt style of notebook 4.2, ToolBench part): the question and the 50
  candidates numbered in their retriever's order, each shown as its title and its first 60 words; the model returns
  the numbers of the 5 passages most needed to answer, most useful first (structured output). If it returns fewer
  than 5 valid numbers, the list is filled from the candidate order. The 5 chosen passages, in full, go to the
  existing answer prompt; EM and F1 are the existing code.
- **Graph (original)** — the cached graph answers above, unchanged.

**Metric.** Token F1 per question (as in the primary result); EM and recall@5 of the gold supporting paragraphs are
secondary, as is recall@50 of each candidate list.

**Contrasts** (paired bootstrap over questions, 5,000 resamples, seed 7, `paired_boot`; per benchmark):
1. **graph − (hybrid + rerank)**: does a stronger baseline close the graph's gain?
2. **(graph + rerank) − (hybrid + rerank)**: does the graph still add something once both sides are reranked?

These 4 intervals (2 contrasts × 2 benchmarks) join the paper's family of headline intervals for the Bonferroni check
(28 + 4 = 32 intervals, factor z(0.05/64)/z(0.025)); the widened interval is printed next to the plain one.

**Decision rule** (`baselines.md`): if neither contrast changes notebook 1.4a's verdict on either benchmark,
notebooks 1.5, 4.1 and 4.5 keep their current retriever.

**Cost.** Two Batch jobs: 4,000 rerank calls (about 4k input tokens each) and 4,000 answer calls; estimated about
$1 at Batch prices. Stages `nb11_rerank` and `nb11_ans_rerank`.

```python
# Hybrid candidates: BM25 (pure numpy/scipy, Okapi k1=1.5, b=0.75) + dense, reciprocal rank fusion, top 50.
N_CAND, RRF_K = 50, 60
TOK = re.compile(r"\w+")

def bm25_index(texts, k1=1.5, b=0.75):
    vocab, rows, cols, vals = {}, [], [], []
    for i, t in enumerate(texts):
        for w, c in Counter(TOK.findall(t.lower())).items():
            rows.append(i); cols.append(vocab.setdefault(w, len(vocab))); vals.append(c)
    tf = sp.csr_matrix((np.array(vals, np.float64), (rows, cols)), shape=(len(texts), len(vocab)))
    dl = np.asarray(tf.sum(1)).ravel(); avgdl = dl.mean()
    df = np.bincount(tf.indices, minlength=len(vocab))
    idf = np.log(1 + (len(texts) - df + 0.5) / (df + 0.5))
    tf = tf.tocoo()
    w = idf[tf.col] * tf.data * (k1 + 1) / (tf.data + k1 * (1 - b + b * dl[tf.row] / avgdl))
    return vocab, sp.csc_matrix((w, (tf.row, tf.col)), shape=tf.shape)

def bm25_scores(index, query):
    vocab, W = index
    ids = [vocab[w] for w in TOK.findall(query.lower()) if w in vocab]   # repeated query terms count again
    return np.asarray(W[:, ids].sum(1)).ravel() if ids else np.zeros(W.shape[0])

def rrf(*scores):
    fused = np.zeros(len(scores[0]))
    for s in scores:
        rank = np.empty(len(s), int); rank[np.argsort(-s, kind="stable")] = np.arange(len(s))
        fused += 1 / (RRF_K + rank + 1)
    return fused

def graph_ranked(d, qi, k=N_CAND):
    # The graph retriever of `retrieve` above, unchanged, returning its top k passages instead of its top K.
    q = d["Q"][qi]
    dense = d["P"] @ q
    pers = np.zeros(d["T"].shape[0], np.float64)
    fs = d["F"] @ q if len(d["facts"]) else np.zeros(0)
    for j in np.argsort(-fs)[:5]:
        a, b, pi, _ = d["facts"][j]
        pers[a] += fs[j]; pers[b] += fs[j]
    for j in np.argsort(-dense)[:20]:
        pers[j] += 0.05 * max(dense[j], 0)
    if pers.sum() == 0:                              # same fallback as `retrieve`: the dense order
        return list(np.argsort(-dense)[:k]), 0
    x = ppr(d["T"], pers)[:d["n_p"]]
    return list(np.argsort(-x)[:k]), int((x > 0).sum())

t0 = time.time()
for name, d in DATA.items():
    bm = bm25_index(d["passages"])
    d["cand"] = []
    n_pos = []
    for qi, q in enumerate(d["questions"]):
        dense = d["P"] @ d["Q"][qi]
        hyb = [int(i) for i in np.argsort(-rrf(bm25_scores(bm, q["question"]), dense), kind="stable")[:N_CAND]]
        g50, npos = graph_ranked(d, qi)
        g50 = [int(i) for i in g50]
        assert g50[:K] == [int(i) for i in d["ret"][qi][1]], (name, qi)   # same graph retriever as above
        d["cand"].append({"vec": hyb, "graph": g50}); n_pos.append(npos)
    for m in ("vec", "graph"):
        r5 = np.nanmean([recall(d, q, c[m][:K]) for q, c in zip(d["questions"], d["cand"])])
        r50 = np.nanmean([recall(d, q, c[m]) for q, c in zip(d["questions"], d["cand"])])
        print(f"{name} {'hybrid' if m == 'vec' else 'graph':<6} candidates: recall@5 before rerank {r5:.1%} | "
              f"recall@50 {r50:.1%}")
    print(f"{name}: graph candidates with a positive PageRank score: min {min(n_pos):,}, median "
          f"{int(np.median(n_pos)):,} (so the graph always yields a full list of {N_CAND})")
print(f"candidate lists in {time.time() - t0:.0f} s")
```

```python
# The same LLM reranker for both candidate lists (prompt style of notebook 4.2), then the existing answer prompt.
class Ranked(BaseModel):
    ranked: list[int] = Field(description="numbers of the 5 passages most needed to answer the question, most useful first")

RERANK_SYS = ("You select passages for a multi-hop question. You receive the question and a numbered list of candidate "
              "passages, each shown as its title and its opening words. Return the numbers of the 5 passages that are "
              "most needed to answer the question, the most useful first. A multi-hop question needs several passages: "
              "include a passage for every step of the reasoning chain.")

def snippet(p, n_words=60):
    title, _, text = p.partition("\n")
    words = text.split()
    return f"{title}: {' '.join(words[:n_words])}{' ...' if len(words) > n_words else ''}"

def rerank_prompt(d, qi, cands):
    lines = [f"[{n}] {snippet(d['passages'][i])}" for n, i in enumerate(cands, 1)]
    return "Candidate passages:\n" + "\n".join(lines) + f"\n\nQuestion: {d['questions'][qi]['question']}"

RR_SYS = {"vec": "hyb", "graph": "graphrr"}          # rerank input -> name of the reranked system
reqs = {}
for name, d in DATA.items():
    for qi, q in enumerate(d["questions"]):
        for m in RR_SYS:
            reqs[f"{name}|{m}|{q['qid']}"] = LLM.body([{"role": "system", "content": RERANK_SYS},
                                                       {"role": "user", "content": rerank_prompt(d, qi, d["cand"][qi][m])}],
                                                      max_tokens=100, schema=Ranked)
RR = llm.run("nb11_rerank", reqs, parse=Ranked)
print(f"{len(reqs):,} rerank calls, {sum(r['parsed'] is None for r in RR.values())} unparsed")

def reranked(name, m, qi):
    cands = DATA[name]["cand"][qi][m]
    p = RR[f"{name}|{m}|{DATA[name]['questions'][qi]['qid']}"]["parsed"] or {"ranked": []}
    chosen = list(dict.fromkeys(cands[n - 1] for n in p["ranked"] if isinstance(n, int) and 1 <= n <= len(cands)))
    return (chosen + [j for j in cands if j not in chosen])[:K], min(len(chosen), K)

reqs = {}
for name, d in DATA.items():
    d["rr"] = []
    for qi, q in enumerate(d["questions"]):
        row = {}
        for m, s in RR_SYS.items():
            ids, n_sel = reranked(name, m, qi)
            row[s] = (ids, n_sel)
            reqs[f"{name}|{s}|{q['qid']}"] = LLM.body([{"role": "system", "content": ANSWER_SYS},
                                                       {"role": "user", "content": f"{ctx(d, ids)}\n\nQuestion: {q['question']}"}],
                                                      max_tokens=300)
        d["rr"].append(row)
ANS_RR = llm.run("nb11_ans_rerank", reqs)
print(f"{len(reqs):,} answers")

for name, d in DATA.items():
    for qi, q in enumerate(d["questions"]):
        for s, (ids, n_sel) in d["rr"][qi].items():
            pred = final(ANS_RR[f"{name}|{s}|{q['qid']}"]["text"])
            d["scores"][qi][s] = {"pred": pred, "em": float(max(normalize(pred) == normalize(g) for g in q["golds"])),
                                  "f1": max(f1(pred, g) for g in q["golds"]), "recall": recall(d, q, ids),
                                  "n_sel": n_sel}
    short = sum(d["rr"][qi][s][1] < K for qi in range(len(d["questions"])) for s in RR_SYS.values())
    print(f"{name}: {short} of {2 * len(d['questions'])} rerank outputs gave fewer than {K} valid numbers (filled from the candidate order)")
```

```python
# Results of the amendment: the two contrasts per benchmark, with a Bonferroni-widened interval as a check.
from scipy.stats import norm
from smash_batch import PRICE
BONF = norm.ppf(1 - 0.05 / 64) / norm.ppf(0.975)    # 28 headline intervals + these 4 = 32, two-sided
LABEL = {"vector": "vector (top-5 dense)", "graph": "graph (original)", "hyb": "hybrid + rerank", "graphrr": "graph + rerank"}
CONTRASTS = [("graph", "hyb", "graph - (hybrid + rerank)"), ("graphrr", "hyb", "(graph + rerank) - (hybrid + rerank)"),
             ("graph", "vector", "graph - vector (original primary, for reference)")]

RR_SUMMARY = {}
for name, d in DATA.items():
    S = d["scores"]; n = len(S)
    print(f"\n{name}  (n = {n})")
    print(f"{'system':<24}{'EM [95% CI]':>22}{'F1':>8}{'recall@5':>10}")
    for s in ("vector", "graph", "hyb", "graphrr"):
        k = int(sum(r[s]["em"] for r in S)); lo, hi = wilson(k, n)
        print(f"{LABEL[s]:<24}{f'{k / n:.1%} [{lo:.0%}-{hi:.0%}]':>22}{np.mean([r[s]['f1'] for r in S]):>8.1%}"
              f"{np.nanmean([r[s]['recall'] for r in S]):>10.1%}")
    print(f"{'F1 contrast':<52}{'mean [95% CI]':>26}{'Bonferroni-widened':>24}")
    for a, b, lab in CONTRASTS:
        dm, dlo, dhi = paired_boot([r[b]["f1"] for r in S], [r[a]["f1"] for r in S])
        wlo, whi = dm - BONF * (dm - dlo), dm + BONF * (dhi - dm)
        RR_SUMMARY[(name, a, b)] = (dm, dlo, dhi, wlo, whi)
        print(f"{lab:<52}{f'{dm:+.1%} [{dlo:+.1%}, {dhi:+.1%}]':>26}{f'[{wlo:+.1%}, {whi:+.1%}]':>24}")
    if name == "musique":
        print("By hop count, (graph + rerank) - (hybrid + rerank):",
              "; ".join(f"{g} {paired_boot([r['hyb']['f1'] for r in S if r['group'] == g], [r['graphrr']['f1'] for r in S if r['group'] == g])[0]:+.1%}"
                        for g in sorted({r['group'] for r in S})))

print("\nTokens and cost of the amendment (both benchmarks):")
for stage, pre, lab in (("nb11_rerank", "|vec|", "rerank, hybrid candidates"), ("nb11_rerank", "|graph|", "rerank, graph candidates"),
                        ("nb11_ans_rerank", "|hyb|", "answers, hybrid + rerank"), ("nb11_ans_rerank", "|graphrr|", "answers, graph + rerank")):
    rows = [r["r"] for k, r in llm.rows(stage).items() if pre in k and k.split("|")[0] in DATA and k.split("|")[2] in
            {q["qid"] for q in DATA[k.split("|")[0]]["questions"]}]
    i, o = sum(r.get("in", 0) for r in rows), sum(r.get("out", 0) for r in rows)
    usd = sum((r.get("in", 0) * PRICE[r.get("tier", "batch")][0] + r.get("out", 0) * PRICE[r.get("tier", "batch")][1]) / 1e6 for r in rows)
    print(f"  {lab:<28}{len(rows):>6,} calls {i / max(len(rows), 1):>7,.0f} in / {o / max(len(rows), 1):>4,.0f} out tokens per call  ${usd:.2f}")

print("\nVerdict (sensitivity analysis; the original primary result above stands):")
for name in DATA:
    v = lambda t: "graph better" if t[1] > 0 else "graph worse" if t[2] < 0 else "no resolvable difference"
    print(f"  {name}: original graph - vector: {verdicts[name]}; graph - (hybrid + rerank): {v(RR_SUMMARY[(name, 'graph', 'hyb')])}; "
          f"(graph + rerank) - (hybrid + rerank): {v(RR_SUMMARY[(name, 'graphrr', 'hyb')])}")
```

Recall@5 of the supporting passages: MuSiQue 55.3 / 58.4 / 70.6 / 76.2; 2Wiki 70.6 / 84.4 / 80.7 / 94.4 (same order).
With the Bonferroni widening (32 intervals), graph − (hybrid + rerank) is not resolvable on either dataset, and
(graph + rerank) − (hybrid + rerank) stays resolvable on both ([+2.3, +10.2] and [+5.6, +14.3]).

**Verdict** (sensitivity analysis; the original primary result stands): graph better than dense top-5 retrieval
(primary). Against the stronger hybrid + rerank retriever, the unreranked graph is no longer resolvably better on
MuSiQue (−2.3) and only narrowly on 2Wiki (+3.2). **Once both sides use the same reranker, the graph adds +6.3 F1 on
MuSiQue and +9.9 on 2Wiki**, because its 50 candidates contain more of the supporting passages (recall@50 86.5% vs
80.3%, 94.8% vs 80.8%). By hops on MuSiQue: +8.9 at 2 hops, +5.6 at 3, −0.7 at 4. The graph's value here is as a
candidate generator for a reranker; the gain vanishes at four hops, as in the original result.
